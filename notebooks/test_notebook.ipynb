## numpy

```python
rng = np.random.default_rng(0)
a = rng.normal(size=(200, 200))
print(np.__version__, round(float(np.linalg.norm(a @ a.T) / 1e4), 3))
```

## scipy

```python
import scipy
from scipy import ndimage
from scipy.optimize import minimize
```

```python
res = minimize(lambda x: (x[0] - 3) ** 2 + (x[1] + 1) ** 2, x0=[0.0, 0.0])
lab, n = ndimage.label(np.eye(5) + np.eye(5)[::-1] > 0)
print(scipy.__version__, np.round(res.x, 3), n)  # [3, -1] 9 (4-connected X)
```

## scikit-image

```python
import skimage
from skimage.draw import polygon
from skimage.measure import approximate_polygon, find_contours
```

```python
img = np.zeros((60, 60))
rr, cc = polygon([10, 30, 50, 30], [30, 50, 30, 10], img.shape)
img[rr, cc] = 1
contour = find_contours(img, 0.5)[0]
approx = approximate_polygon(contour, tolerance=1.5)
print(skimage.__version__, int(img.sum()), len(approx) - 1)  # a diamond: 4 corners
```

## matplotlib

```python
import matplotlib
import matplotlib.pyplot as plt
```

```python
fig, ax = plt.subplots(figsize=(4, 3))
x = np.linspace(0, 2 * np.pi, 100)
ax.plot(x, np.sin(x))
ax.set_title("matplotlib")
print(matplotlib.__version__)
plt.show()
```
